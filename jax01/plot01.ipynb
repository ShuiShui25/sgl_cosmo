```python
import os
import numpy as np
import matplotlib.pyplot as plt

# pip install corner
import corner

npz_label = "w0waCDM"
npz_path = f"/home/geng/Codes/lensing_cosmo/jax01/output/posterior_minimal_chunks_{npz_label}.npz"
outdir   = os.path.dirname(npz_path)

data = np.load(npz_path)

# 你的 keep_keys（脚本里固定的顺序）
keys = [
    "Om", "w",
    "gamma0", "gamma_s", "log_sig_g",
    "delta0", "delta_s", "log_sig_d",
    "beta0", "log_sig_b",
]

# 堆成 (Nsamples, 10)
samples = np.vstack([data[k] for k in keys]).T
print("samples shape =", samples.shape)

# -------------------------
# 1) trace plots
# -------------------------
fig, axes = plt.subplots(len(keys), 1, figsize=(10, 2.0 * len(keys)), sharex=True)
for i, k in enumerate(keys):
    axes[i].plot(samples[:, i], linewidth=0.6)
    axes[i].set_ylabel(k)
axes[-1].set_xlabel("sample index")
fig.tight_layout()
trace_path = os.path.join(outdir, f"trace_10params_{npz_label}.png")
fig.savefig(trace_path, dpi=200)
plt.close(fig)
print("Saved:", trace_path)

# -------------------------
# 2) corner plot
# -------------------------
# 可选：丢掉前 20% 作为 burn-in（因为你把 warmup+chunk1 samples 也存进来了）
burn = int(0.6 * samples.shape[0])
samples2 = samples[burn:, :]

fig = corner.corner(
    samples2,
    labels=keys,
    show_titles=True,
    title_fmt=".3f",
    quantiles=[0.16, 0.5, 0.84],
)
corner_path = os.path.join(outdir, f"corner_10params_{npz_label}.png")
fig.savefig(corner_path, dpi=200)
plt.close(fig)
print("Saved:", corner_path)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# pip install getdist
from getdist import MCSamples, plots as gdplots

plot_label = "wcdm-tri"
npz_path = f"/Users/bwu127/Desktop/geng/out /posterior_minimal_chunks_{plot_label}.npz"
outdir   = os.path.dirname(npz_path)

data = np.load(npz_path)

# -------------------------
# 0) keys auto-select
# -------------------------
tail_keys = [
    "gamma0", "gamma_s", "log_sig_g",
    "delta0", "delta_s", "log_sig_d",
    "beta0", "log_sig_b",
]

pl = plot_label.lower()
if pl.startswith("wcdm"):
    head_keys = ["Om", "w"]
elif pl.startswith("w0wa"):
    head_keys = ["Om", "w0", "wa"]
elif pl.startswith("wphi"):
    head_keys = ["Om", "w0", "alpha"]
else:
    raise ValueError(
        f"plot_label='{plot_label}' 无法识别前缀。请用 wcdm*/w0wa*/wphi* 之一。"
    )

keys = head_keys + tail_keys

# 检查 npz 里是否都有对应字段
missing = [k for k in keys if k not in data.files]
if missing:
    raise KeyError(f"npz 文件缺少这些 keys: {missing}\n当前可用: {data.files}")

# 堆成 (Nsamples, Ndim)
samples = np.vstack([data[k] for k in keys]).T
print("samples shape =", samples.shape)

# -------------------------
# 1) trace plots (仍用 matplotlib，getdist 不负责 trace)
# -------------------------
fig, axes = plt.subplots(len(keys), 1, figsize=(10, 2.0 * len(keys)), sharex=True)
for i, k in enumerate(keys):
    axes[i].plot(samples[:, i], linewidth=0.6)
    axes[i].set_ylabel(k)
axes[-1].set_xlabel("sample index")
fig.tight_layout()
trace_path = os.path.join(outdir, f"trace_{len(keys)}params_{plot_label}.png")
fig.savefig(trace_path, dpi=200)
plt.close(fig)
print("Saved:", trace_path)

# -------------------------
# 2) getdist triangle plot
# -------------------------
burn = int(0.2 * samples.shape[0])
samples2 = samples[burn:, :]

# 你也可以在这里做 thinning（可选）
# thin = 1
# samples2 = samples2[::thin]

# getdist 的 name 建议用简单标识符（不要太花哨）
# label 用来显示（可用 LaTeX）
name_map = {
    "Om": "Om", "w": "w", "w0": "w0", "wa": "wa", "alpha": "alpha",
    "gamma0": "gamma0", "gamma_s": "gamma_s", "log_sig_g": "log_sig_g",
    "delta0": "delta0", "delta_s": "delta_s", "log_sig_d": "log_sig_d",
    "beta0": "beta0", "log_sig_b": "log_sig_b",
}
label_map = {
    "Om": r"\Omega_m",
    "w": r"w",
    "w0": r"w_0",
    "wa": r"w_a",
    "alpha": r"\alpha",
    "gamma0": r"\gamma_0",
    "gamma_s": r"\gamma_s",
    "log_sig_g": r"\log\sigma_\gamma",
    "delta0": r"\delta_0",
    "delta_s": r"\delta_s",
    "log_sig_d": r"\log\sigma_\delta",
    "beta0": r"\beta_0",
    "log_sig_b": r"\log\sigma_\beta",
}

names  = [name_map[k] for k in keys]
labels = [label_map.get(k, k) for k in keys]

mc = MCSamples(
    samples=samples2,
    names=names,
    labels=labels,
    label=plot_label,
)

g = gdplots.get_subplot_plotter(width_inch=10)
# triangle_plot 会同时画 1D 边缘分布 + 2D 等高线
g.triangle_plot(
    [mc],
    filled=True,
    title_limit=1,      # 标题显示范围（防太长）
)

triangle_path = os.path.join(outdir, f"getdist_triangle_{len(keys)}params_{plot_label}.png")
g.export(triangle_path)
print("Saved:", triangle_path)
```
