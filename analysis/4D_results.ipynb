```python
import os
import numpy as np
import matplotlib.pyplot as plt
from getdist import plots, MCSamples

def calculate_marginal_median_and_mad(samples):
    # Flatten (nwalkers, nsteps, ndim) → (nwalkers*nsteps, ndim)
    nwalkers, nsteps, ndim = samples.shape
    flat = samples.reshape(-1, ndim)
    med = np.median(flat, axis=0)
    mad = np.median(np.abs(flat - med), axis=0)
    return med, mad

def plot_triangle_one(samples, param_names, label,
                      output_folder, plot_name, marker=True):
    """
    Triangle plot for a single MCMC chain.

    samples:    numpy array shape (nwalkers, nsteps, ndim)
    param_names: list of 4 latex strings, e.g. [r'$\Omega_m$',r'$w_0$',r'$w_a$',r'$\beta$']
    label:      string for legend
    """
    # wrap into a GetDist MCSamples object
    mcs = MCSamples(samples=samples, names=param_names, label=label)

    # corner‐plot setup
    g = plots.getSubplotPlotter(width_inch=8)
    g.settings.title_limit_fontsize = 14
    g.settings.axes_fontsize = 14
    g.settings.axes_labelsize = 14

    # make the triangle plot (filled contours)
    g.triangle_plot(
        [mcs], param_names,
        filled=True,
        contour_colors=['#4169E1'],
        contour_lws=[1.5],
        alpha=0.6
    )

    if marker:
        # add vertical dashed lines at the median
        medians, mads = calculate_marginal_median_and_mad(samples)
        for i, x0 in enumerate(medians):
            ax = g.subplots[i, i]
            ax.axvline(x0, color='k', ls='--')
            # overwrite the default title with “param = median ± MAD”
            ax.set_title(f"{param_names[i]} = {medians[i]:.3f}±{mads[i]:.3f}", 
                         fontsize=12)

    # save
    os.makedirs(output_folder, exist_ok=True)
    plt.savefig(os.path.join(output_folder, plot_name), 
                bbox_inches='tight', dpi=200)
    plt.close()

# ———————————————— usage ——————————————————

# 1) Four parameter names in LaTeX
param_names = [r'$\Omega_m$', r'$w_0$', r'$w_a$', r'$\beta$']

# 2) Load your single chain
# samples_file = os.path.join(
#     os.path.dirname(__file__),  # script directory → portable
#     'Output', 
#     'mcmc_chain.npy'
# )
samples_file = r"/home/gengs/Codes/projects/GLS_gamma_cosmology/GLS_cosmo01/Output/CPL-good-Tri/mcmc_cosmop4D-CPL-good-Tri.npy"
samples = np.load(samples_file)  # shape (nwalkers, nsteps, ndim)

# 3) Call the plot routine
plot_triangle_one(
    samples=samples,
    param_names=param_names,
    label='MyRun',
    output_folder=os.path.join(os.path.dirname(__file__),'Output'),
    plot_name='triangle_4param.png',
    marker=True
)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from getdist import plots, MCSamples

def get_base_dir():
    # In a script, __file__ exists; in notebooks it doesn't
    if '__file__' in globals():
        return os.path.dirname(os.path.abspath(__file__))
    else:
        return os.getcwd()

def calculate_marginal_median_and_mad(samples):
    nwalkers, nsteps, ndim = samples.shape
    flat = samples.reshape(-1, ndim)
    med = np.median(flat, axis=0)
    mad = np.median(np.abs(flat - med), axis=0)
    return med, mad

def plot_triangle_one(samples, param_names, label,
                      output_folder, plot_name, marker=True):
    mcs = MCSamples(samples=samples, names=param_names, label=label)
    g = plots.getSubplotPlotter(width_inch=8)
    g.settings.title_limit_fontsize = 14
    g.settings.axes_fontsize = 14
    g.settings.axes_labelsize = 14

    g.triangle_plot(
        [mcs], param_names,
        filled=True,
        contour_colors=['#4169E1'],
        contour_lws=[1.5],
        alpha=0.6
    )

    if marker:
        medians, mads = calculate_marginal_median_and_mad(samples)
        for i, x0 in enumerate(medians):
            ax = g.subplots[i, i]
            ax.axvline(x0, color='k', ls='--')
            ax.set_title(f"{param_names[i]} = {medians[i]:.3f}±{mads[i]:.3f}",
                         fontsize=12)

    os.makedirs(output_folder, exist_ok=True)
    plt.savefig(os.path.join(output_folder, plot_name),
                bbox_inches='tight', dpi=200)
    plt.show()
    

# ———————————————— usage ——————————————————
base_dir = get_base_dir()

# 1) Four parameter names in LaTeX
param_names = [r'$\Omega_m$', r'$w_0$', r'$w_a$', r'$\beta$']

# 2) Load your single chain from relative path
# samples_file = os.path.join(base_dir, 'Output', 'mcmc_chain.npy')
samples_file = r"/home/gengs/Codes/projects/GLS_gamma_cosmology/GLS_cosmo01/Output/CPL-good-Tri/mcmc_cosmop4D-CPL-good-Tri.npy"
samples = np.load(samples_file)  # shape (nwalkers, nsteps, ndim)

# 3) Call the plot routine with relative output folder
plot_triangle_one(
    samples=samples,
    param_names=param_names,
    label='MyRun',
    output_folder=os.path.join(base_dir, 'Output'),
    plot_name='triangle_4param.png',
    marker=True
)
```
